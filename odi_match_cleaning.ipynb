```python
df.columns = df.columns.str.strip()

print("Column names cleaned successfully!")
print(df.columns.tolist())
```

```python
text_columns = df.select_dtypes(
    include=["object", "string"]
).columns

for col in text_columns:
    df[col] = df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Text values cleaned successfully!")
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values[missing_values > 0])
```

```python
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).round(2)

missing_report = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": missing_percentage
})

missing_report = missing_report[
    missing_report["Missing Values"] > 0
]

missing_report
```

```python
extras_columns = [
    "wides",
    "noballs",
    "byes",
    "legbyes",
    "penalty"
]

for col in extras_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        ).fillna(0)

print("Missing numeric extras handled!")
```

```python
wicket_columns = [
    "wicket_type",
    "player_dismissed"
]

for col in wicket_columns:
    if col in df.columns:
        df[col] = df[col].replace(
            r"^\s*$",
            np.nan,
            regex=True
        )

print("Wicket information checked!")
```

```python
empty_columns = [
    col for col in df.columns
    if df[col].isna().all()
]

print("Completely empty columns:")
print(empty_columns)
```

```python
df = df.drop(columns=empty_columns)

print("Completely empty columns removed!")
print("Updated dataset shape:", df.shape)
```

```python
print(df.dtypes)
```

```python
integer_columns = [
    "match_id",
    "innings",
    "runs_off_bat",
    "extras",
    "wides",
    "noballs",
    "byes",
    "legbyes",
    "penalty"
]

for col in integer_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        ).astype("Int64")
```

```python
df["start_date"] = pd.to_datetime(
    df["start_date"],
    errors="coerce"
)
```

```python
df["ball"] = df["ball"].astype("string")
```

```python
run_columns = [
    "runs_off_bat",
    "extras",
    "wides",
    "noballs",
    "byes",
    "legbyes",
    "penalty"
]

for col in run_columns:
    if col in df.columns:
        print(
            col,
            "minimum =", df[col].min(),
            "maximum =", df[col].max()
        )
```

```python
for col in run_columns:
    if col in df.columns:
        invalid = df[df[col] < 0]

        print(
            f"{col}: {len(invalid)} negative values"
        )
```

```python
df["total_runs"] = (
    df["runs_off_bat"].fillna(0)
    + df["extras"].fillna(0)
)
```
